```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\Lenovo\Downloads\archive (26).zip")
```

```python
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.columns
```

```python
df.isnull().sum()
```

```python
df['Acronym'] = df['Acronym'].fillna('Unknown')
```

```python
df['Interventions'] = df['Interventions'].fillna('Not Mentioned')
```

```python
df['Outcome Measures'].fillna('Not Available')
```

```python
df['Gender'] = df['Gender'].fillna('Unknown')
```

```python
df['Phases'] = df['Phases'].fillna('Not Specified')
```

```python
df['Enrollment'] = df['Enrollment'].fillna(df['Enrollment'].median())
```

```python
df['Other IDs'] = df['Other IDs'].fillna('Not Available')
```

```python
df = df.dropna(subset=['Start Date'])
```

```python
df['Primary Completion Date'] = df['Primary Completion Date'].fillna('Unknown')
```

```python
df['Results First Posted'] = df['Results First Posted'].fillna('Not Posted')
```

```python
df['Locations'] = df['Locations'].fillna('Unknown')
```

```python
df['Study Documents'] = df['Study Documents'].fillna('No Documents')
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
df['Status'].value_counts().plot(kind='bar')
plt.title("Clinical Trial Status")
plt.xlabel("Status")
plt.ylabel("Count")
plt.show()
```

```python
df['Study Type'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Study Type Distribution")
plt.ylabel('')
plt.show()
```

```python
df['Phases'].value_counts().head(10).plot(kind='bar')
plt.title("Top Trial Phases")
plt.show()
```

```python
df['Conditions'].value_counts().head(10).plot(kind='bar')
plt.title("Top Conditions")
plt.show()
```

```python
df['Interventions'].value_counts().head(10).plot(kind='bar')
plt.title("Top Interventions")
plt.show()
```

```python
df['Enrollment'].describe()
```

```python
df['Start Date'] = pd.to_datetime(df['Start Date'],errors='coerce')
df['Completion Date'] = pd.to_datetime(df['Completion Date'],errors='coerce')
```

```python
df['Trial Duration'] = (df['Completion Date'] - df['Start Date']).dt.days
```

```python
df['Trial Duration'].describe()
```

```python
sns.histplot(df['Trial Duration'])
plt.title("Trial Duration Distribution")
plt.xlabel("Days")
plt.show()
```

```python
df['Start Year'] = df['Start Date'].dt.year
```

```python
df['Start Year'].value_counts().sort_index().plot(kind='line')
plt.title("Trials Started Per Year")
plt.xlabel("Years")
plt.ylabel("Number of Trials")
plt.show()
```

```python
df['Start Month'] = df['Start Date'].dt.month
```

```python
df['Start Month'].value_counts().sort_index().plot(kind='bar')
plt.title("Trials Started by Month")
plt.xlabel("Month")
plt.ylabel("Count")
plt.show()
```

```python
df['Enrollment Category'] = pd.cut(df['Enrollment'],bins=[0,100,1000,10000,100000],
                                   labels=['Small','Medium','Large','Very Large'])
```

```python
df['Enrollment Category'].value_counts().plot(kind='bar')
plt.title("Enrollment Category")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

```python
df['Is Completed'] = df['Status'].apply(lambda x:1 if x=='Completed' else 0 )
```

```python
df['Is Completed'].value_counts().plot(kind='pie',autopct='%1.1f%%')
plt.title("Completed vs Non Completed Trials")
plt.ylabel('')
plt.show()
```

```python
df['Multi Country Trial'] = df['Locations'].apply(lambda x:1 if',' in str(x) else 0)
```

```python
df['Multi Country Trial'].value_counts().plot(kind='bar')
plt.title("Multi Country Trials")
plt.xlabel("0=No , 1=Yes")
plt.ylabel("Count")
plt.show()
```

```python
df['Number of Interventions'] = df['Interventions'].apply(lambda x: len(str(x).split('|')))
```

```python
df['Result Available'] = df['Results First Posted'].apply(lambda x: 0 if x=='Not Posted' else 1)
```

```python
df['Result Available'].value_counts().plot(kind='pie', autopct='%1.1f%%')
plt.title("Result Availability")
plt.ylabel('')
plt.show()
```
